# ATVED - Unified Traffic Violations Model Training

This notebook trains a YOLOv8 model to detect **Helmets**, **No Helmets**, and **Triple Riding** all at once using a Roboflow Universe dataset.

### Step 1: Ensure you are using a GPU
In Colab, go to **Runtime > Change runtime type** and select **T4 GPU**.

```python
!pip install ultralytics roboflow
from ultralytics import YOLO
from roboflow import Roboflow
import torch

print(f"CUDA available: {torch.cuda.is_available()}")
```

### Step 2: Download the Dataset from Roboflow
Go to Roboflow Universe and find a dataset (see the suggestions provided). Click **Download Dataset**, select **YOLOv8** format, and choose **Show Download Code**. Paste your API key below.

```python
# PASTE YOUR ROBOFLOW DOWNLOAD CODE HERE!
# It will look something like this:
rf = Roboflow(api_key="YOUR_ROBOFLOW_API_KEY_HERE")
project = rf.workspace("WORKSPACE_NAME").project("PROJECT_NAME")
version = project.version(1)
dataset = version.download("yolov8")
```

### Step 3: Train the Unified YOLOv8 Model
We will fine-tune the `yolov8s.pt` (small) model. It strikes the best balance between high accuracy for tiny objects (like helmets far away) and fast real-time inference speed.

```python
model = YOLO('yolov8s.pt') # Load a pretrained YOLOv8 Small model

# Train the model!
# dataset.location contains the path to the downloaded data.yaml file
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,       # 50 epochs is usually the sweet spot for Roboflow datasets
    imgsz=640,       # Standard image size
    batch=16,        # Maximize GPU memory usage
    project="ATVED_Unified",
    name="traffic_violations_model"
)
```

### Step 4: Validate and Export
Once training is done, your best model weights will be saved in `ATVED_Unified/traffic_violations_model/weights/best.pt`. Download this file and drop it into your VS Code project folder!
