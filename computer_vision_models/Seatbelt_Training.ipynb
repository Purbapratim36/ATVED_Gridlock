## 2. Download Dataset from Roboflow
Go to Roboflow Universe, find a Seatbelt Dataset, click `Download Dataset` (YOLOv8 format), and paste the code snippet below:

```python
from roboflow import Roboflow

# ==========================================
# PASTE YOUR ROBOFLOW SNIPPET EXACTLY HERE:
# ==========================================
rf = Roboflow(api_key="YOUR_API_KEY_HERE")
project = rf.workspace("WORKSPACE").project("PROJECT")
version = project.version(1)
dataset = version.download("yolov8")
```

## 3. Train the YOLOv8 Model
We use `yolov8n.pt` as the base model for rapid learning.

```python
from ultralytics import YOLO

# Load a pretrained YOLO model
model = YOLO('yolov8n.pt')

# Train the model
results = model.train(
    data=f"{dataset.location}/data.yaml", 
    epochs=30,  # Increase to 50 or 100 for better accuracy later
    imgsz=640, 
    batch=16,
    plots=True
)
```
