## Step 3: Train YOLOv8 🚀

We start from pre-trained YOLOv8n weights and fine-tune on our license plate data.

````
from ultralytics import YOLO

# Load pre-trained YOLOv8 nano model
model = YOLO("yolov8n.pt")

# Train
results = model.train(
    data=data_yaml_path,
    epochs=30,           # 30 epochs is usually enough for plates
    imgsz=640,
    batch=16,
    name="atved_plate", 
    save=True,
    verbose=True,
)

print("\n  ✅ TRAINING COMPLETE!")
````
